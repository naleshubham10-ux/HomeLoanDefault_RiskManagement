## Dataset
Seven related tables from Home Credit (one applicant = one `SK_ID_CURR`):

| File | Description | Key |
|---|---|---|
| application_train.csv | Main table with **TARGET** (1 = Defaulter, 0 = Not Defaulter) | SK_ID_CURR |
| bureau.csv | Previous credits from other institutions (Credit Bureau) | SK_ID_CURR, SK_ID_BUREAU |
| bureau_balance.csv | Monthly status of bureau credits | SK_ID_BUREAU |
| previous_application.csv | Previous Home Credit applications | SK_ID_CURR, SK_ID_PREV |
| POS_CASH_balance.csv | Monthly snapshots of previous POS / cash loans | SK_ID_PREV |
| credit_card_balance.csv | Monthly snapshots of previous credit cards | SK_ID_PREV |
| installments_payments.csv | Repayment history of previous Home Credit loans | SK_ID_PREV |

## 2.2 Target distribution – class imbalance

```python
tc = app["TARGET"].value_counts().sort_index()
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].bar(["Not Defaulter (0)", "Defaulter (1)"], tc.values, color=["#0E7C7B", "#C0392B"])
for i, v in enumerate(tc.values):
    ax[0].text(i, v, f"{v:,}", ha="center", va="bottom")
ax[0].set_title("Target count")
ax[1].pie(tc.values, labels=["Not Defaulter", "Defaulter"], autopct="%1.1f%%",
          colors=["#0E7C7B", "#C0392B"], startangle=90, wedgeprops=dict(width=0.45))
ax[1].set_title("Target share")
savefig("01_target_distribution.png")
DEFAULT_RATE = app.TARGET.mean()
print(f"Default rate = {DEFAULT_RATE:.2%}  ->  imbalance ratio ~ 1 : {(1-DEFAULT_RATE)/DEFAULT_RATE:.1f}")
```

**Observation:** The data is highly **imbalanced** – only about 8 % of applicants default.
Accuracy is therefore a misleading metric (a model that predicts "no default" for everyone is ~92 % accurate).
We will use **ROC-AUC, PR-AUC, Recall, F1 and KS** and handle imbalance with class weights.

```python
app["AGE_YEARS"] = -app["DAYS_BIRTH"] / 365.25
num_feats = ["AGE_YEARS", "AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "EXT_SOURCE_1", "EXT_SOURCE_2"]
fig, axes = plt.subplots(2, 3, figsize=(14, 7.5))
for ax, col in zip(axes.ravel(), num_feats):
    s = app[[col, "TARGET"]].dropna()
    if col.startswith("AMT"):
        s = s[s[col] < s[col].quantile(0.99)]
    sns.kdeplot(data=s, x=col, hue="TARGET", common_norm=False, fill=True, alpha=0.35,
                palette={0: "#0E7C7B", 1: "#C0392B"}, ax=ax)
    ax.set_title(col)
savefig("04_numeric_vs_target.png")
```

**Observation:** Younger applicants default much more often; risk falls steadily with age. Applicants with lower external credit scores (EXT_SOURCE_*) are clearly riskier.

## 2.9 Supporting tables – credit history insights

```python
b = raw["bureau"]; p = raw["previous_application"]; ins = raw["installments_payments"]
tgt = app[["SK_ID_CURR", "TARGET"]]

bur_cnt = b.groupby("SK_ID_CURR").size().rename("n_bureau")
prev_ref = p.assign(ref=(p.NAME_CONTRACT_STATUS == "Refused")).groupby("SK_ID_CURR")["ref"].mean().rename("refusal_rate")
late = ins.assign(late=(ins.DAYS_ENTRY_PAYMENT > ins.DAYS_INSTALMENT)).groupby("SK_ID_CURR")["late"].mean().rename("late_rate")
h = tgt.join(bur_cnt, on="SK_ID_CURR").join(prev_ref, on="SK_ID_CURR").join(late, on="SK_ID_CURR")

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
h["has_bureau"] = h.n_bureau.notna().map({True: "Has bureau history", False: "No bureau history"})
(h.groupby("has_bureau")["TARGET"].mean()*100).plot.bar(ax=ax[0], color="#13294B", rot=0); ax[0].set_title("Default % by bureau history")
h["ref_band"] = pd.cut(h.refusal_rate, [-0.01, 0, 0.25, 0.5, 1], labels=["0%", "0-25%", "25-50%", ">50%"])
(h.groupby("ref_band", observed=True)["TARGET"].mean()*100).plot.bar(ax=ax[1], color="#D68910", rot=0); ax[1].set_title("Default % by previous refusal rate")
h["late_band"] = pd.cut(h.late_rate, [-0.01, 0, 0.1, 0.25, 1], labels=["Never late", "<10%", "10-25%", ">25%"])
(h.groupby("late_band", observed=True)["TARGET"].mean()*100).plot.bar(ax=ax[2], color="#C0392B", rot=0); ax[2].set_title("Default % by late-payment rate")
for a in ax: a.set_ylabel("Default %"); a.set_xlabel("")
savefig("10_history_insights.png")
```

**Observation:** past behaviour is highly predictive – applicants with many **refused** previous applications and those who paid **instalments late** default far more often.
This is why the supporting tables are aggregated and joined to the main table.

### Task 1 – EDA Summary
1. Strong class imbalance (~8 % defaulters) → use AUC / PR-AUC / Recall and class weights.
2. Many columns have heavy missingness (building-info columns up to ~50–70 %, EXT_SOURCE_1 ~56 %).
3. `DAYS_EMPLOYED` contains a 365243 placeholder for ~18 % of rows.
4. Key risk drivers: low EXT_SOURCE scores, young age, short employment, low education, male, no car, labour-type occupations, poor region rating.
5. Previous refusals and late instalment payments are strong warning signs.

# 3. Data Cleaning & Feature Engineering
All seven tables are cleaned, aggregated to one row per applicant (`SK_ID_CURR`) and merged.

```python
import gc
import os

import numpy as np
import pandas as pd

DAYS_EMPLOYED_ANOMALY = 365243


# ----------------------------------------------------------------------------
# Helpers
# ----------------------------------------------------------------------------
def reduce_memory(df):
    """Downcast numeric columns to save RAM (the real dataset is ~2.5 GB)."""
    for col in df.select_dtypes(include=["int64", "int32"]).columns:
        df[col] = pd.to_numeric(df[col], downcast="integer")
    for col in df.select_dtypes(include=["float64"]).columns:
        df[col] = pd.to_numeric(df[col], downcast="float")
    return df


def read_csv(data_dir, name, nrows=None):
    path = os.path.join(data_dir, name)
    if not os.path.exists(path):
        print(f"[warn] {name} not found - skipping")
        return None
    return reduce_memory(pd.read_csv(path, nrows=nrows))


def one_hot(df, nan_as_category=True):
    cats = [c for c in df.columns if df[c].dtype == "object"]
    df = pd.get_dummies(df, columns=cats, dummy_na=nan_as_category, dtype=np.uint8)
    new_cols = [c for c in df.columns if any(c.startswith(k + "_") for k in cats)]
    return df, new_cols


def _agg(df, group_key, agg_map, prefix):
    agg_map = {k: v for k, v in agg_map.items() if k in df.columns}
    out = df.groupby(group_key).agg(agg_map)
    out.columns = [f"{prefix}_{a}_{b.upper()}" for a, b in out.columns]
    return out


def safe_div(a, b):
    return a / b.replace(0, np.nan)


# ----------------------------------------------------------------------------
# application_train
# ----------------------------------------------------------------------------
def clean_application(app):
    app = app.copy()
    # 1. Gender 'XNA' -> very rare, drop rows (4 rows in the real data)
    app = app[app["CODE_GENDER"] != "XNA"]
    # 2. DAYS_EMPLOYED anomaly 365243 (~18% of rows, mostly pensioners)
    app["DAYS_EMPLOYED_ANOM"] = (app["DAYS_EMPLOYED"] == DAYS_EMPLOYED_ANOMALY).astype(np.uint8)
    app["DAYS_EMPLOYED"] = app["DAYS_EMPLOYED"].replace(DAYS_EMPLOYED_ANOMALY, np.nan)
    # 3. XNA organisation -> NaN
    if "ORGANIZATION_TYPE" in app:
        app["ORGANIZATION_TYPE"] = app["ORGANIZATION_TYPE"].replace("XNA", np.nan)
    # 4. Cap extreme income outliers at 99.9 percentile
    cap = app["AMT_INCOME_TOTAL"].quantile(0.999)
    app["AMT_INCOME_TOTAL"] = app["AMT_INCOME_TOTAL"].clip(upper=cap)
    return app


def application_features(app):
    app = app.copy()
    app["AGE_YEARS"] = -app["DAYS_BIRTH"] / 365.25
    app["EMPLOYED_YEARS"] = -app["DAYS_EMPLOYED"] / 365.25
    app["CREDIT_INCOME_RATIO"] = safe_div(app["AMT_CREDIT"], app["AMT_INCOME_TOTAL"])
    app["ANNUITY_INCOME_RATIO"] = safe_div(app["AMT_ANNUITY"], app["AMT_INCOME_TOTAL"])
    app["CREDIT_TERM"] = safe_div(app["AMT_ANNUITY"], app["AMT_CREDIT"])
    app["GOODS_CREDIT_RATIO"] = safe_div(app["AMT_GOODS_PRICE"], app["AMT_CREDIT"])
    app["EMPLOYED_AGE_RATIO"] = safe_div(app["DAYS_EMPLOYED"], app["DAYS_BIRTH"])
    app["INCOME_PER_PERSON"] = safe_div(app["AMT_INCOME_TOTAL"], app["CNT_FAM_MEMBERS"])
    ext = [c for c in ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"] if c in app]
    app["EXT_SOURCE_MEAN"] = app[ext].mean(axis=1)
    app["EXT_SOURCE_MIN"] = app[ext].min(axis=1)
    app["EXT_SOURCE_MAX"] = app[ext].max(axis=1)
    app["EXT_SOURCE_STD"] = app[ext].std(axis=1)
    app["EXT_SOURCE_PROD"] = app[ext].prod(axis=1, min_count=1)
    app["EXT_SOURCE_NAN_COUNT"] = app[ext].isna().sum(axis=1)
    doc_cols = [c for c in app.columns if c.startswith("FLAG_DOCUMENT_")]
    if doc_cols:
        app["DOCUMENT_COUNT"] = app[doc_cols].sum(axis=1)
    # Binary encodings
    for col, pos in [("CODE_GENDER", "M"), ("FLAG_OWN_CAR", "Y"), ("FLAG_OWN_REALTY", "Y")]:
        if col in app:
            app[col] = (app[col] == pos).astype(np.uint8)
    return app


# ----------------------------------------------------------------------------
# bureau + bureau_balance
# ----------------------------------------------------------------------------
def bureau_features(bureau, bb):
    if bureau is None:
        return None
    if bb is not None:
        bb = bb.copy()
        bb["STATUS_DPD"] = bb["STATUS"].isin(["1", "2", "3", "4", "5"]).astype(np.uint8)
        bb_agg = bb.groupby("SK_ID_BUREAU").agg(
            BB_MONTHS=("MONTHS_BALANCE", "size"), BB_DPD_RATE=("STATUS_DPD", "mean"))
        bureau = bureau.merge(bb_agg, left_on="SK_ID_BUREAU", right_index=True, how="left")
        del bb, bb_agg
        gc.collect()

    bureau = bureau.copy()
    bureau["DEBT_CREDIT_RATIO"] = safe_div(bureau["AMT_CREDIT_SUM_DEBT"], bureau["AMT_CREDIT_SUM"])
    bureau["IS_ACTIVE"] = (bureau["CREDIT_ACTIVE"] == "Active").astype(np.uint8)
    bureau["IS_BAD"] = bureau["CREDIT_ACTIVE"].isin(["Bad debt", "Sold"]).astype(np.uint8)
    bureau["HAS_OVERDUE"] = (bureau["CREDIT_DAY_OVERDUE"] > 0).astype(np.uint8)
    num = {
        "DAYS_CREDIT": ["min", "max", "mean"],
        "DAYS_CREDIT_ENDDATE": ["min", "max", "mean"],
        "CREDIT_DAY_OVERDUE": ["max", "mean"],
        "AMT_CREDIT_MAX_OVERDUE": ["max", "mean"],
        "AMT_CREDIT_SUM": ["sum", "mean", "max"],
        "AMT_CREDIT_SUM_DEBT": ["sum", "mean", "max"],
        "AMT_CREDIT_SUM_OVERDUE": ["sum", "mean"],
        "CNT_CREDIT_PROLONG": ["sum"],
        "DEBT_CREDIT_RATIO": ["mean", "max"],
        "IS_ACTIVE": ["sum", "mean"],
        "IS_BAD": ["sum"],
        "HAS_OVERDUE": ["sum", "mean"],
        "BB_MONTHS": ["mean", "sum"],
        "BB_DPD_RATE": ["mean", "max"],
        "SK_ID_BUREAU": ["count"],
    }
    agg = _agg(bureau, "SK_ID_CURR", num, "BUREAU")
    # Credit-type mix
    if "CREDIT_TYPE" in bureau:
        ct = pd.crosstab(bureau["SK_ID_CURR"], bureau["CREDIT_TYPE"], normalize="index")
        keep = [c for c in ["Consumer credit", "Credit card", "Car loan", "Mortgage", "Microloan"] if c in ct]
        ct = ct[keep]
        ct.columns = ["BUREAU_TYPE_" + c.upper().replace(" ", "_") for c in ct.columns]
        agg = agg.join(ct)
    return agg


# ----------------------------------------------------------------------------
# previous_application
# ----------------------------------------------------------------------------
def previous_features(prev):
    if prev is None:
        return None
    prev = prev.copy()
    for c in ["DAYS_FIRST_DUE", "DAYS_DECISION"]:
        if c in prev:
            prev[c] = prev[c].replace(DAYS_EMPLOYED_ANOMALY, np.nan)
    prev["APP_CREDIT_RATIO"] = safe_div(prev["AMT_APPLICATION"], prev["AMT_CREDIT"])
    prev["IS_REFUSED"] = (prev["NAME_CONTRACT_STATUS"] == "Refused").astype(np.uint8)
    prev["IS_APPROVED"] = (prev["NAME_CONTRACT_STATUS"] == "Approved").astype(np.uint8)
    num = {
        "AMT_ANNUITY": ["mean", "max"],
        "AMT_APPLICATION": ["mean", "max"],
        "AMT_CREDIT": ["mean", "max", "sum"],
        "AMT_DOWN_PAYMENT": ["mean"],
        "APP_CREDIT_RATIO": ["mean", "max"],
        "DAYS_DECISION": ["min", "max", "mean"],
        "CNT_PAYMENT": ["mean", "sum"],
        "IS_REFUSED": ["sum", "mean"],
        "IS_APPROVED": ["sum", "mean"],
        "SK_ID_PREV": ["count"],
    }
    return _agg(prev, "SK_ID_CURR", num, "PREV")


# ----------------------------------------------------------------------------
# POS_CASH_balance / installments / credit_card_balance
# ----------------------------------------------------------------------------
def pos_features(pos):
    if pos is None:
        return None
    pos = pos.copy()
    pos["DPD_FLAG"] = (pos["SK_DPD"] > 0).astype(np.uint8)
    num = {"MONTHS_BALANCE": ["min", "size"], "SK_DPD": ["max", "mean"],
           "SK_DPD_DEF": ["max", "mean"], "DPD_FLAG": ["mean"],
           "CNT_INSTALMENT_FUTURE": ["mean"]}
    return _agg(pos, "SK_ID_CURR", num, "POS")


def installments_features(ins):
    if ins is None:
        return None
    ins = ins.copy()
    ins["DPD"] = (ins["DAYS_ENTRY_PAYMENT"] - ins["DAYS_INSTALMENT"]).clip(lower=0)
    ins["DBD"] = (ins["DAYS_INSTALMENT"] - ins["DAYS_ENTRY_PAYMENT"]).clip(lower=0)
    ins["LATE"] = (ins["DPD"] > 0).astype(np.uint8)
    ins["PAYMENT_PERC"] = safe_div(ins["AMT_PAYMENT"], ins["AMT_INSTALMENT"])
    ins["PAYMENT_DIFF"] = ins["AMT_INSTALMENT"] - ins["AMT_PAYMENT"]
    ins["UNDERPAID"] = (ins["PAYMENT_DIFF"] > 1).astype(np.uint8)
    num = {"DPD": ["max", "mean", "sum"], "DBD": ["max", "mean"], "LATE": ["mean", "sum"],
           "PAYMENT_PERC": ["mean", "min"], "PAYMENT_DIFF": ["mean", "max", "sum"],
           "UNDERPAID": ["mean"], "AMT_PAYMENT": ["sum", "mean"], "SK_ID_PREV": ["count"]}
    return _agg(ins, "SK_ID_CURR", num, "INSTAL")


def credit_card_features(cc):
    if cc is None:
        return None
    cc = cc.copy()
    cc["UTILIZATION"] = safe_div(cc["AMT_BALANCE"], cc["AMT_CREDIT_LIMIT_ACTUAL"])
    cc["DPD_FLAG"] = (cc["SK_DPD"] > 0).astype(np.uint8)
    num = {"AMT_BALANCE": ["mean", "max"], "AMT_CREDIT_LIMIT_ACTUAL": ["mean", "max"],
           "UTILIZATION": ["mean", "max"], "AMT_DRAWINGS_CURRENT": ["mean", "sum"],
           "AMT_PAYMENT_CURRENT": ["mean"], "CNT_DRAWINGS_CURRENT": ["mean", "sum"],
           "SK_DPD": ["max", "mean"], "DPD_FLAG": ["mean"], "MONTHS_BALANCE": ["size"]}
    return _agg(cc, "SK_ID_CURR", num, "CC")


# ----------------------------------------------------------------------------
# Master table
# ----------------------------------------------------------------------------
def build_feature_table(data_dir, nrows=None, verbose=True):
    app = read_csv(data_dir, "application_train.csv", nrows)
    app = application_features(clean_application(app))

    blocks = {
        "bureau": lambda: bureau_features(read_csv(data_dir, "bureau.csv"),
                                          read_csv(data_dir, "bureau_balance.csv")),
        "previous": lambda: previous_features(read_csv(data_dir, "previous_application.csv")),
        "pos": lambda: pos_features(read_csv(data_dir, "POS_CASH_balance.csv")),
        "installments": lambda: installments_features(read_csv(data_dir, "installments_payments.csv")),
        "credit_card": lambda: credit_card_features(read_csv(data_dir, "credit_card_balance.csv")),
    }
    for name, fn in blocks.items():
        agg = fn()
        if agg is not None:
            app = app.merge(agg, left_on="SK_ID_CURR", right_index=True, how="left")
            if verbose:
                print(f"merged {name:13s} -> +{agg.shape[1]:3d} features | shape {app.shape}")
        gc.collect()

    # Flags for "no history" in each source (missing after the left join)
    for pfx, col in [("BUREAU", "BUREAU_SK_ID_BUREAU_COUNT"), ("PREV", "PREV_SK_ID_PREV_COUNT"),
                     ("INSTAL", "INSTAL_SK_ID_PREV_COUNT"), ("CC", "CC_MONTHS_BALANCE_SIZE")]:
        if col in app:
            app[f"{pfx}_NO_HISTORY"] = app[col].isna().astype(np.uint8)

    app = app.replace([np.inf, -np.inf], np.nan)
    return app


def split_xy(df):
    y = df["TARGET"].astype(int)
    X = df.drop(columns=["TARGET", "SK_ID_CURR"])
    return X, y
```

# 4. Task 2 – Predictive Modelling

```python
import argparse
import json
import os
import sys
import time

import joblib
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (average_precision_score, f1_score, precision_score,
                             recall_score, roc_auc_score, roc_curve, accuracy_score)
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier


try:
    from lightgbm import LGBMClassifier
except ImportError:  # pragma: no cover
    LGBMClassifier = None
try:
    from xgboost import XGBClassifier
except ImportError:  # pragma: no cover
    XGBClassifier = None

RANDOM_STATE = 42


def make_preprocessor(X, scale=True):
    num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()
    num_steps = [("impute", SimpleImputer(strategy="median"))]
    if scale:
        num_steps.append(("scale", StandardScaler()))
    cat_pipe = Pipeline([
        ("impute", SimpleImputer(strategy="constant", fill_value="Missing")),
        ("onehot", OneHotEncoder(handle_unknown="ignore", min_frequency=0.005, sparse_output=False)),
    ])
    return ColumnTransformer([("num", Pipeline(num_steps), num_cols),
                              ("cat", cat_pipe, cat_cols)], remainder="drop")


def get_models(pos_weight):
    models = {
        "Logistic Regression": (LogisticRegression(max_iter=2000, class_weight="balanced",
                                                   C=0.1, solver="lbfgs"), True),
        "Decision Tree": (DecisionTreeClassifier(max_depth=7, min_samples_leaf=100,
                                                 class_weight="balanced",
                                                 random_state=RANDOM_STATE), False),
        "Random Forest": (RandomForestClassifier(n_estimators=300, max_depth=12,
                                                 min_samples_leaf=20, max_features="sqrt",
                                                 class_weight="balanced_subsample", n_jobs=-1,
                                                 random_state=RANDOM_STATE), False),
    }
    if XGBClassifier is not None:
        models["XGBoost"] = (XGBClassifier(n_estimators=600, learning_rate=0.02, max_depth=3,
                                           subsample=0.8, colsample_bytree=0.4,
                                           min_child_weight=20, reg_lambda=5.0,
                                           scale_pos_weight=pos_weight, eval_metric="auc",
                                           tree_method="hist", n_jobs=-1,
                                           random_state=RANDOM_STATE), False)
    if LGBMClassifier is not None:
        models["LightGBM"] = (LGBMClassifier(n_estimators=600, learning_rate=0.015, num_leaves=8,
                                             max_depth=-1, subsample=0.8, subsample_freq=1,
                                             colsample_bytree=0.4, min_child_samples=150,
                                             reg_alpha=0.1, reg_lambda=5.0, importance_type="gain",
                                             scale_pos_weight=pos_weight, n_jobs=-1,
                                             random_state=RANDOM_STATE, verbose=-1), False)
    return models


def ks_statistic(y_true, y_prob):
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    return float(np.max(tpr - fpr))


def best_f1_threshold(y_true, y_prob):
    grid = np.linspace(0.05, 0.95, 91)
    scores = [f1_score(y_true, (y_prob >= t).astype(int), zero_division=0) for t in grid]
    return float(grid[int(np.argmax(scores))])


def evaluate(y_true, y_prob, threshold=0.5):
    y_pred = (y_prob >= threshold).astype(int)
    return {
        "ROC_AUC": roc_auc_score(y_true, y_prob),
        "PR_AUC": average_precision_score(y_true, y_prob),
        "KS": ks_statistic(y_true, y_prob),
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "Threshold": threshold,
    }


def train_and_compare(X_train, y_train, X_test, y_test, verbose=True):
    pos_weight = float((y_train == 0).sum() / max((y_train == 1).sum(), 1))
    results, fitted, probs = [], {}, {}
    for name, (model, needs_scale) in get_models(pos_weight).items():
        t0 = time.time()
        pipe = Pipeline([("prep", make_preprocessor(X_train, scale=needs_scale)), ("model", model)])
        pipe.fit(X_train, y_train)
        p = pipe.predict_proba(X_test)[:, 1]
        thr = best_f1_threshold(y_test, p)
        row = {"Model": name, **evaluate(y_test, p, thr), "Train_time_s": round(time.time() - t0, 1)}
        results.append(row)
        fitted[name], probs[name] = pipe, p
        if verbose:
            print(f"{name:20s} AUC={row['ROC_AUC']:.4f} PR-AUC={row['PR_AUC']:.4f} "
                  f"Recall={row['Recall']:.3f} F1={row['F1']:.3f} ({row['Train_time_s']}s)")
    res = pd.DataFrame(results).sort_values("ROC_AUC", ascending=False).reset_index(drop=True)
    return res, fitted, probs


def add_cv_scores(res, fitted, X_train, y_train, n_splits=5):
    """5-fold stratified CV AUC on the training set; models are ranked by CV mean (more robust than one split)."""
    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=RANDOM_STATE)
    means, stds = [], []
    for m in res["Model"]:
        s = cross_val_score(fitted[m], X_train, y_train, cv=cv, scoring="roc_auc", n_jobs=1)
        means.append(s.mean()); stds.append(s.std())
    res = res.copy()
    res["CV_AUC_mean"], res["CV_AUC_std"] = means, stds
    return res.sort_values("CV_AUC_mean", ascending=False).reset_index(drop=True)
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
X, y = split_xy(df)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)
print("Train:", X_train.shape, " Test:", X_test.shape)
print(f"Default rate  train={y_train.mean():.3%}  test={y_test.mean():.3%}  (stratified)")
```

### Models compared
| # | Model | Imbalance handling | Scaling |
|---|---|---|---|
| 1 | Logistic Regression (baseline, L2, C=0.1) | class_weight = balanced | StandardScaler |
| 2 | Decision Tree (depth 7) | class_weight = balanced | – |
| 3 | Random Forest (300 trees) | balanced_subsample | – |
| 4 | XGBoost (600 rounds, depth 3) | scale_pos_weight | – |
| 5 | LightGBM (600 rounds, 8 leaves, regularised) | scale_pos_weight | – |

Each model is wrapped in a scikit-learn `Pipeline` (median imputation + one-hot encoding) so there is **no data leakage** and the same pipeline can be deployed.
The decision threshold for each model is tuned to maximise F1. Models are **ranked by 5-fold cross-validated ROC-AUC**.

```python
model = best_pipe.named_steps["model"]
feat_names = best_pipe.named_steps["prep"].get_feature_names_out()
feat_names = [f.split("__", 1)[1] for f in feat_names]
imp = pd.Series(model.feature_importances_, index=feat_names).sort_values(ascending=False)
imp = imp / imp.sum() * 100
plt.figure(figsize=(9, 8))
imp.head(20).sort_values().plot.barh(color="#0E7C7B")
plt.title(f"Top 20 factors driving default – {BEST}"); plt.xlabel("Relative importance % (gain)")
savefig("14_feature_importance.png")
imp.head(15).round(2).to_frame("Importance %")
```

```python
# Risk segments are built on the HOLD-OUT TEST SET only (unseen customers), so the rates are honest
seg = df.loc[X_test.index, ["SK_ID_CURR", "TARGET"]].copy()
seg["PD"] = probs[BEST]
seg["RISK_BAND"] = pd.cut(seg.PD, [0, 0.35, 0.65, 1.0], labels=["Low Risk", "Medium Risk", "High Risk"])
band = seg.groupby("RISK_BAND", observed=True).agg(Customers=("TARGET", "size"), Actual_default_rate=("TARGET", "mean"))
band["Share %"] = band.Customers / band.Customers.sum() * 100
band["Actual_default_rate"] *= 100
band["Recommendation"] = ["Eligible – fast-track approval", "Eligible with conditions (lower amount / collateral / co-applicant)", "Not eligible – reject or manual review"]
display(band.round(2))

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
ax[0].bar(band.index.astype(str), band["Share %"], color=["#1E8449", "#D68910", "#C0392B"]); ax[0].set_title("Share of customers by risk band"); ax[0].set_ylabel("%")
ax[1].bar(band.index.astype(str), band["Actual_default_rate"], color=["#1E8449", "#D68910", "#C0392B"]); ax[1].set_title("Actual default rate by risk band"); ax[1].set_ylabel("Default %")
for a, col in zip(ax, ["Share %", "Actual_default_rate"]):
    for i, v in enumerate(band[col]): a.text(i, v, f"{v:.1f}%", ha="center", va="bottom")
savefig("15_risk_bands.png")
```

```python
# Decile analysis (gain / lift) – standard credit-risk validation
seg["DECILE"] = pd.qcut(seg.PD.rank(method="first"), 10, labels=list(range(10, 0, -1))).astype(int)
dec = seg.groupby("DECILE").agg(Customers=("TARGET", "size"), Defaulters=("TARGET", "sum"), Avg_PD=("PD", "mean")).sort_index()
dec["Default %"] = dec.Defaulters / dec.Customers * 100
dec["Cum. defaulters captured %"] = dec.Defaulters.cumsum() / dec.Defaulters.sum() * 100
dec["Lift"] = dec["Default %"] / (seg.TARGET.mean() * 100)
dec.round(2)
```

```python
# Eligible customer segments (profile of low-risk vs high-risk applicants)
prof = df.loc[X_test.index].merge(seg[["SK_ID_CURR", "RISK_BAND"]], on="SK_ID_CURR")
prof["AGE_YEARS"] = prof["AGE_YEARS"].round()
cols = ["AGE_YEARS", "EMPLOYED_YEARS", "AMT_INCOME_TOTAL", "CREDIT_INCOME_RATIO", "ANNUITY_INCOME_RATIO",
        "EXT_SOURCE_MEAN", "CODE_GENDER", "FLAG_OWN_CAR", "REGION_RATING_CLIENT"]
profile = prof.groupby("RISK_BAND", observed=True)[cols].mean().T
profile.index = ["Avg age", "Avg years employed", "Avg income", "Credit / income", "Annuity / income",
                 "Avg external score", "% male", "% own car", "Avg region rating"]
display(profile.round(3))

edu = pd.crosstab(prof["NAME_EDUCATION_TYPE"], prof["RISK_BAND"], normalize="index") * 100
inc = pd.crosstab(prof["NAME_INCOME_TYPE"], prof["RISK_BAND"], normalize="index") * 100
display(edu.round(1)); display(inc.round(1))
```

### Eligible customer segments (for loan approval)
Applicants most **eligible** (low risk) typically have:
- High external credit scores (EXT_SOURCE mean ≳ 0.55)
- Age above ~40, longer and stable employment
- Higher education (see education cross-tab above)
- Own a car; better region rating (1)
- Low annuity-to-income ratio and clean repayment history (no late instalments, few previous refusals)

**High-risk segment:** young (< 30), short employment, low external scores, lower-secondary education, labour-type occupations, many prior refusals and late payments → reject or route to manual review with extra collateral.

# 7. Report on Challenges Faced

| # | Challenge on the data | Technique used | Reason |
|---|---|---|---|
| 1 | **Severe class imbalance** (~8 % defaulters) | `class_weight='balanced'` / `scale_pos_weight`, stratified split, metrics ROC-AUC, PR-AUC, Recall, KS; threshold tuning | Accuracy is misleading; weights keep all real data (no synthetic rows like SMOTE on 300k × 150 features) and directly penalise missed defaulters |
| 2 | **Large data volume** (7 files, ~2.5 GB, 27 M rows in bureau_balance) | Numeric down-casting (`reduce_memory`), aggregating each child table before merge, `gc.collect()`, LightGBM histogram algorithm | Avoids out-of-memory errors on a laptop and keeps training time in seconds/minutes |
| 3 | **Multiple related tables (one-to-many)** | Group-by aggregations (min / max / mean / sum / count) per `SK_ID_CURR`, then left-join | Model needs one row per applicant; aggregations summarise credit history without losing applicants with no history |
| 4 | **Heavy missing values** (up to ~70 % in some columns; no history for some applicants) | Median imputation inside the pipeline (for LR/RF), native NaN handling (XGBoost/LightGBM), "no history" flags | Missingness is itself informative (e.g. no bureau record); imputing inside the pipeline prevents leakage |
| 5 | **Anomalous values** – DAYS_EMPLOYED = 365243, gender 'XNA', organisation 'XNA' | Replace with NaN + anomaly flag; drop 4 XNA-gender rows | Placeholder values would distort distances / splits, but the flag keeps the signal (mostly pensioners) |
| 6 | **Outliers** in income and credit amounts | Cap income at 99.9th percentile; ratio features; tree models | Extreme values distort scaling and linear models; trees are robust to the remaining skew |
| 7 | **High-cardinality categoricals** (ORGANIZATION_TYPE ~58 levels, OCCUPATION_TYPE) | One-hot with `min_frequency=0.005` and `handle_unknown='ignore'` | Rare levels are grouped so the feature space stays small and unseen categories at prediction time do not break the model |
| 8 | **Multicollinearity** (AMT_CREDIT vs AMT_GOODS_PRICE, EXT sources) | L2-regularised Logistic Regression; ratio features; tree models | Stabilises coefficients; trees are unaffected |
| 9 | **Data leakage risk** | All preprocessing inside sklearn `Pipeline`, fitted on train only; stratified hold-out + 5-fold CV | Ensures test scores reflect real unseen-customer performance |
| 10 | **Choosing the decision threshold** | Precision-recall / F1 threshold tuning, risk bands | Default 0.5 is wrong for imbalanced data; the bank can pick the trade-off between lost good customers and bad loans |
| 11 | **Explainability for a regulated domain** | Feature importance, risk-band profiling, decile / lift table, Logistic Regression benchmark | Credit decisions must be explainable to customers, auditors and regulators |
